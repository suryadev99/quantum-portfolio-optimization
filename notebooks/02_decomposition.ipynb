# 02 – Inside the decomposition
Clusters, picks per cluster, and the QUBO of one cluster.

```python
from qportfolio.config.loader import load_config
from qportfolio.pipeline import run_pipeline
import numpy as np

res = run_pipeline(load_config('../configs/demo_small.yaml'))
print(res.cluster_summary)
print('picks per cluster:', res.allocation.base.tolist())
```

```python
key, qubo = next(iter(res.qubos.items()))
print('cluster, k =', key, '| qubits =', qubo.num_qubits, '| penalty =', qubo.penalty, qubo.info)
pool = res.pools[key]
print(len(pool), 'candidates; best local objective', pool.objective[0])
```
